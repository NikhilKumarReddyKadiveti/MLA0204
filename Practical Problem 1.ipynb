```python
import pandas as pd

def find_s_algorithm(training_data):
    """
    Implements the FIND-S algorithm to find the most specific hypothesis.

    Args:
        training_data (list of lists): A list of training examples, where each
                                      example is a list of attribute values
                                      followed by the classification (e.g., 'Yes' or 'No').

    Returns:
        list: The most specific hypothesis.
    """
    # Initialize the most specific hypothesis with the most specific possible values ('0')
    # The number of attributes is len(training_data[0]) - 1 (excluding the classification)
    num_attributes = len(training_data[0]) - 1
    hypothesis = ['0'] * num_attributes

    print("\nInitialization of Hypothesis:", hypothesis)

    for i, example in enumerate(training_data):
        attributes = example[:-1] # All but the last element are attributes
        target = example[-1]     # The last element is the target classification

        if target == 'Yes':
            print(f"\nProcessing Positive Example {i+1}: {example}")
            # If it's the first positive example, initialize hypothesis directly
            if hypothesis[0] == '0':
                hypothesis = list(attributes)
                print("   Hypothesis updated to first positive example:", hypothesis)
            else:
                # Generalize hypothesis for each attribute
                for j in range(num_attributes):
                    if hypothesis[j] != attributes[j]:
                        hypothesis[j] = '?' # Generalize if attributes differ
                print("   Hypothesis after generalization:", hypothesis)
        else:
            print(f"\nIgnoring Negative Example {i+1}: {example}")

    return hypothesis

# --- Demonstration ---

# Sample training data (e.g., Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport)
# Note: FIND-S only considers positive examples.
# Data is represented as a list of lists.
# The last element of each inner list is the target classification ('Yes' or 'No').
training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

print("Training Data:")
for row in training_data:
    print(row)

# Run the FIND-S algorithm
most_specific_hypothesis = find_s_algorithm(training_data)

print("\n--- Result ---")
print("The most specific hypothesis is:", most_specific_hypothesis)

# You can also use pandas for a more structured view (optional)
df = pd.DataFrame(training_data, columns=['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport'])
print("\nTraining Data (DataFrame):")
print(df)
```
